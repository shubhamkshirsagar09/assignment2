```python
# Rule-Based Student Grading Expert System

print("Student Grading Expert System")

# Input marks for five subjects
s1 = float(input("Enter marks for Subject 1: "))
s2 = float(input("Enter marks for Subject 2: "))
s3 = float(input("Enter marks for Subject 3: "))
s4 = float(input("Enter marks for Subject 4: "))
s5 = float(input("Enter marks for Subject 5: "))

# Calculate total and percentage
total = s1 + s2 + s3 + s4 + s5
percentage = total / 5

# Display total and percentage
print("\nTotal Marks:", total, "/ 500")
print("Percentage:", percentage, "%")

# Rule-based grading using if-else
if percentage < 40:
    print("Grade: Fail")
elif percentage < 65:
    print("Grade: II Class")
elif percentage < 75:
    print("Grade: I Class")
else:
    print("Grade: Distinction")
```
